```python
from omegaconf import OmegaConf
from team_gm.models.structure_flow import StructureFlowClient

config = OmegaConf.load("configs/baseline.yaml")
config = StructureFlowClient.Config(**config)
client = StructureFlowClient(config)
```

```python
import pickle
from team_gm.data.dataloader import ProteinData
from team_gm.utils.data_utils import write_pdb


with open(client.config.data.data_pkl_path, "rb") as f:
    pdb_ids, pdb_weights, train_pdb, valid_pdb = pickle.load(f)

train_dataset = ProteinData(
    item_dict=train_pdb,
    input_dir_path=config.data.input_dir_path
)

valid_dataset = ProteinData(
    item_dict=valid_pdb,
    input_dir_path=config.data.input_dir_path
)
```

```python
batch = valid_dataset.__getitem__(0)
write_pdb(
    atom_pos=batch.atom_pos,
    atom_mask=batch.atom_mask,
    res_names=batch.res_name,
    seq_idx=batch.seq_idx,
)
```
